# Technical Selection Rationale

## Final Anomaly Detection Method

The Tuned Isolation Forest was frozen as the final anomaly detection method.

The method was selected after systematic hyperparameter experiments on Day 19. Validation anomaly-rate stability was used because ground-truth anomaly labels are unavailable. Day 20 feature ablation and temporal monitoring analysis were additionally used to assess robustness and project-specific behaviour.

The model provides anomaly predictions without requiring labelled abnormal observations, which matches the current dataset limitation.

## Final Forecasting Method

The Tuned Random Forest Regressor was frozen as the final forecasting method for next-step pH prediction.

The model was selected through systematic hyperparameter tuning using validation MAE on Day 19. Chronological validation was subsequently performed on Day 20 to examine robustness across different temporal periods.

The test set was not used for model selection or hyperparameter tuning.

## Why These Methods Were Frozen

The final methods provide a practical balance between:

- Validation performance
- Robustness
- Model complexity
- Interpretability of experimental behaviour
- Reproducibility
- Suitability for the available dataset

The anomaly and forecasting tasks are treated separately; therefore, the two models are task-specific final candidates rather than a single overall model.

## Frozen Preprocessing

Missing numerical values are handled using the fitted preprocessing/imputation objects stored with the model artefacts.

The same preprocessing configuration must be applied to future data before inference.

## Frozen Data Split

The project uses chronological train, validation and test partitions.

- Training data → model fitting
- Validation data → tuning and model selection
- Test data → final evaluation only

## Final Evaluation Policy

After this checkpoint, model configuration and preprocessing will not be changed based on test results.

Any future change to the model, features, preprocessing or evaluation methodology will be recorded as a new experiment rather than modifying the frozen configuration.

## Reproducibility Artefacts

The following artefacts have been saved:

- `final_isolation_forest.joblib`
- `final_random_forest.joblib`
- `feature_config.joblib`
- `day22_final_freeze_summary.csv`

## Research Limitations

The current system is a methodological prototype because:

1. The original dataset did not contain timestamps.
2. Timestamps were synthetically generated.
3. Temperature, dissolved oxygen and pressure are unavailable.
4. Ground-truth anomaly labels are unavailable.
5. Forecasting currently focuses on next-step pH.
6. Real underwater sensor data may have different distributions and sampling behaviour.

These limitations must be considered when interpreting the final results.
